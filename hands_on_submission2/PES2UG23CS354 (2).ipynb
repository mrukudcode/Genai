```python
from transformers import pipeline
import ipywidgets as widgets
```

```python
generator = pipeline(
    "text-generation",
    model="gpt2"
)
```

```python
zodiac_dropdown = widgets.Dropdown(
    options=[
        "Aries", "Taurus", "Gemini", "Cancer",
        "Leo", "Virgo", "Libra", "Scorpio",
        "Sagittarius", "Capricorn", "Aquarius", "Pisces"
    ],
    description="Zodiac:"
)

display(zodiac_dropdown)
```

```python
def generate_horoscope(zodiac):
    prompt = f"Today's horoscope for {zodiac}:"

    result = generator(
        prompt,
        max_length=60,
        do_sample=True,
        temperature=0.9
    )

    print(result[0]["generated_text"])
```

```python
generate_button = widgets.Button(description="Generate Horoscope")

def on_button_click(b):
    generate_horoscope(zodiac_dropdown.value)

generate_button.on_click(on_button_click)
display(generate_button)
```
