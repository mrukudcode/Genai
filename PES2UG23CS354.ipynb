```python
from transformers import pipeline
```

```python
# Experiment 1 prompt
gen_prompt = "The future of Artificial Intelligence is"

# Experiment 2 masked sentence
masked_sentence = "The goal of Generative AI is to [MASK] new content."

# Experiment 3 QA inputs
qa_context = "Generative AI poses significant risks such as hallucinations, bias, and deepfakes."
qa_question = "What are the risks?"
```

```python
try:
    bert_gen = pipeline("text-generation", model="bert-base-uncased")
    print(bert_gen(gen_prompt, max_length=30))
except Exception as e:
    print("BERT Generation Failed:", e)
```

```python
try:
    roberta_gen = pipeline("text-generation", model="roberta-base")
    print(roberta_gen(gen_prompt, max_length=30))
except Exception as e:
    print("RoBERTa Generation Failed:", e)
```

```python
bart_gen = pipeline("text-generation", model="facebook/bart-base")
output = bart_gen(gen_prompt, max_length=40, num_return_sequences=1)
print(output[0]["generated_text"])
```

```python
bert_mask = pipeline("fill-mask", model="bert-base-uncased")
bert_preds = bert_mask(masked_sentence)

print("BERT Predictions:")
for p in bert_preds[:5]:
    print(p["token_str"], "->", round(p["score"], 3))
```

```python
roberta_masked_sentence = "The goal of Generative AI is to <mask> new content."

roberta_mask = pipeline("fill-mask", model="roberta-base")
roberta_preds = roberta_mask(roberta_masked_sentence)

print("RoBERTa Predictions:")
for p in roberta_preds[:5]:
    print(p["token_str"], "->", round(p["score"], 3))
```

```python
bart_masked_sentence = "The goal of Generative AI is to <mask> new content."

bart_mask = pipeline("fill-mask", model="facebook/bart-base")
bart_preds = bart_mask(bart_masked_sentence)

print("BART Predictions:")
for p in bart_preds[:5]:
    print(p["token_str"], "->", round(p["score"], 3))
```

```python
file_path = "/content/unit 1.txt"   # change name if needed

with open(file_path, "r", encoding="utf-8") as f:
    context = f.read()

context = context[:1500]  # trim for base models
```

```python
question = "What are the risks?"

models = {
    "BERT": "bert-base-uncased",
    "RoBERTa": "roberta-base",
    "BART": "facebook/bart-base"
}

for name, model in models.items():
    print(f"\n{name} Answer:")
    qa = pipeline("question-answering", model=model)
    result = qa(question=question, context=context)
    print("Answer:", result["answer"])
    print("Score:", round(result["score"], 4))
```
